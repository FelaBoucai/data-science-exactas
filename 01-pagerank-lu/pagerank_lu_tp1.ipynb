### Pregunta 1 - ¿Por que la matriz A definida en (4) es equivalente a $p \ \mathbf{W} \ \mathbf{D} + e \ z^t$ ? Justificar.

Para probar la equivalencia **A** = p **W D** + **e $z^{t}$** podemos analizar cada termino por separado, empezando por **A**.


Cada posicion $a_{ij}$ de la matriz **A** representa la posibilidad de pasar de la pagina j a la página i y se define de esta manera;


$a_{ij}= \left\{ \begin{array}{lcc} (1 − p)/n + (p w_{ij} )/c_{j} & si & c_{j} \neq 0 \\ \\ 1/n & si & c_{j} = 0 \end{array} \right.$

Donde p $\in$ (0,1) es la probabilidad de que el navegante siga un link de la pagina j y $c_{j}$ es 

$c_{j} = \sum_{i=1}^{n} w_{ij}$

La posicion $w_{ij}$ de la matriz **W** es igual a 1 si la página j tiene un link a la página i e igual a 0 en el caso que no tenga ningún link.  

Por otro lado, D es una matriz diagonal de la forma

$d_{jj}= \left\{ \begin{array}{lcc} 1/c_{j} & si & c_{j} \neq 0 \\ \\ 0 & si & c_{j} = 0 \end{array} \right.$

**e** es un vector columna de unos de dimensión n y z es un vector columna cuyos componentes son:

$z_{j}= \left\{ \begin{array}{lcc} (1-p)/n & si & c_{j} \neq 0 \\ \\ 1/n & si & c_{j} = 0 \end{array} \right.$

Una vez explicado cada termino, hay que ver si **A** = p **W D** + **e $z^{t}$** en el caso que $c_{j} \neq 0$ y tambien ver si se cumple cuando $c_{j} = 0$

Primero vemos cuando $c_{j} \neq 0$

**A** = $(1 − p)/n + (p w_{ij} )/c_{j}$

p **W D** = $(p w_{ij} )/c_{j}$

**e $z^{t}$** = (1 − p)/n

Entonces nos queda que p **W D** + **e $z^{t}$** = (1 − p)/n +  $(p w_{ij} )/c_{j}$ = **A**

Ahora cuando $c_{j}$ = 0

**A** = 1/n

p **W D** = 0 

**e $z^{t}$** = 1/n

Sumando nos queda p **W D** + **e $z^{t}$** = 0 + 1/n = **A**

Por lo tanto, queda probado en ambos casos ($c_{j} \neq 0$ y $c_{j} = 0$) que **A** = p **W D** + **e $z^{t}$** 

### Pregunta 2 -  ¿Cómo se garantiza existencia de la factorización $LU$? ¿ La matriz $(\mathbf{I} - p \ \mathbf{W} \ \mathbf{D})$ está bien condicionada? ¿Cómo influye el valor de $p$?

Para garantizar que existe la factorización LU de la matriz vamos a utilizar la proposición 4.2 del apunte de la materia que dice: 

"Si la matriz A es EDD (estrictamente diagonal dominante) entonces, el proceso de eliminación de Gauss no produce pivots nulos."

Entonces, empecemos definiendo la matriz $M = (\mathbf{I} - p \mathbf{W D})$.

Y a nuestra matriz $A = M^{t}$, $A \in \mathbb{R}^{n\times n}$

Ahora vamos a probar, en un principio, (1) que nuestra matriz $A$ tiene descomposición $LU$ por la proposición 4.3, y luego, en una segunda instancia, (2) que dada una matriz $A$, su traspuesta tiene también descomposición $LU$.

(1): Arrancamos probando que nuestra matriz $A\in \mathbb{R}^{n\times n}$, es una matriz EDD:

$A \in \mathbb{R}^{n\times n}$ es EDD $\Longleftrightarrow$ $\sum_{1 \leq j \leq n, j \neq i}^{}|a_{ij}|< |a_{ii}|$ para todo $i:$ $1 \leq i \leq n$

Esto quiere decir, que la suma del módulo todos los elementos de una fila exepto el de la diagonal, son menores que el módulo del elemento de la diagonal.

Por construcción, nuestra matriz $M$ cumple esto pero por columnas. Porque el módulo de los elementos de nuestras columnas (sin contar el de la diagonal) son $1$ divididos por la cantidad de $1$ que hay en esa columna. $\Longrightarrow$ La suma de esos módulos siempre es 1. Esto vale para $p = 1$, pero como $p < 1$, la suma de esos módulos va a ser siempre $<1$.

Ahora, como nuestra martiz $A = M^{t}$. Esto vale por filas, que es lo que necesitamos. $\Longrightarrow A$ es EDD $\Longrightarrow A$ tiene descomposición $LU$ (por la proposición 4.3).

Observación: La matriz $U$ no tiene ceros en la diagonal (pues $det(U) \neq 0$ y $U$ triangular superior).

(2): Ahora queremos ver que que dada una matriz $A$, su traspuesta tiene también descomposición $LU$.

Esto nos va a servir porque $A^{t} = M$, y como $A$ tiene descomposición $LU$, $M$ también la va a tener.

Para esto tomamos que $A^{t} = (LU)^{t} = U^{t}L^{t}$ con $U^{t}$ triangular inferior y $L^{t}$, triangular superior. Pero nos encontramos con el problema de que $U^{t}$ no tiene $1$ en la diagonal. Vamos a solucionar eso:

Para solucionar esto definimos una matriz $S \in \mathbb{R}^{n\times n}$ diagonal, tal que $S_{ii} = U_{ii}$ para todo $1 \leq i \leq n$. Y por lo tanto, $S_{ii}^{-1} = 1/U_{ii}$. ($U_{ii} \neq 0$ por (2))

Bien sabemos que $S^{-1}S = I$ $\Longrightarrow$ $U^{t}L^{t} = U^{t}S^{-1}SL^{t}$.

Ahora, $(U^{t}S^{-1})_{ij} = U_{ij}$ para todo $i \neq j$. Y $(U^{t}S^{-1})_{ij} = U_{ii}^{t} \times S_{ii}^{-1} = U_{ii} \times 1/U_{ii} = 1$ para todo $i = j$.

Llamamos a $\tilde{L} =  U^{t}S^{-1}$ y a $\tilde{U} =  SL^{t}$. Que cumplen $\tilde{L}$ triangular inferior con $1$ en la diagonal y $\tilde{U}$ triangular superior.

Entonces $A^{t} = \tilde{L}\tilde{U}$. Probamos que si $A$ tiene descomposición $LU$ $\Longrightarrow$ $A^{t}$ también tiene.

Para concluir la demostración como $A = M^{t}$ $\Longleftrightarrow$ $A^{t} = M = (\mathbf{I} - p \mathbf{W D})$. Y cómo probamos en (1) que $A$ tiene descomposición $LU$ y en (2) que su traspuesta también. $(\mathbf{I} - p \mathbf{W D})$ tiene descomposición $LU$. 

$\blacksquare$

Con respecto a la segunda pregunta, una matriz está  bien condicionada sí sus $cond_{p}$ tienden a un número cercano al 1, si tienden a un numero considerablemente mayor al 1, se considera mal condicionada. Por lo tanto, hay que analizar la condicion de $(\mathbf{I} - p \mathbf{W D})$.

Sabemos que $cond(\mathbf{I} - p \mathbf{W D}) = ||\mathbf{I} - p \mathbf{W D}|| * ||( \mathbf{I} - p \mathbf{W D})^{-1}||$. Vamos a anlisar cada norma por separado.

$||\mathbf{I} - p \mathbf{W D}|| \leq ||\mathbf{I}|| + || - p \mathbf{W D}||$ 

$||\mathbf{I}|| = 1$

$|| - p \mathbf{W D}|| = |p|*||\mathbf{W D}||$

Para ver $||\mathbf{W D}||$ primero hay que explicar que es $\mathbf{W D}$,

Como ya fue señalado antes, la posicion $w_{ij}$ de la matriz $\mathbf{W}$ es igual a 1 si la página j tiene un link a la página i e igual a 0 en el caso que no tenga ningún link, mientras que $\mathbf{D}$ es una matriz diagonal donde la posicion $d_{jj}$ es igual a 1/$c_j$ si $c_j$ $\neq$ 0 o es igual a 0 en el caso contrario (siendo $c_j$ la cantidad de links que salen de la pagina j). Ya sabiendo esto, se puede decir que las columnas de la matriz $\mathbf{W D}$ estan normalizadas, por lo que la suma de todos los elementos de cada columna siempre va a tener a 1 como maximo (ya que tambien hay casos que es igual a 0), por lo que $||\mathbf{W D}||_1$ = 1

Ahora juntando todo nos queda que

$||\mathbf{I} - p \mathbf{W D}|| \leq 1 + p*1$ 

Falta calcular $||( \mathbf{I} - p \mathbf{W D})^{-1}||$

Para ello podemos usar que $(\mathbf{I} - p \mathbf{W D})^{-1} * (\mathbf{I} - p \mathbf{W D}) = \mathbf{I}$ $\Longleftrightarrow (\mathbf{I} - p \mathbf{W D})^{-1}*\mathbf{I} - (\mathbf{I} - p \mathbf{W D})^{-1}* p \mathbf{W D} = \mathbf{I}$ $\Longleftrightarrow (\mathbf{I} - p \mathbf{W D})^{-1} = \mathbf{I} + (\mathbf{I} - p \mathbf{W D})^{-1}* p \mathbf{W D} $

Utilizando lo obtenido podemos decir que

$||( \mathbf{I} - p \mathbf{W D})^{-1}|| = ||\mathbf{I} + (\mathbf{I} - p \mathbf{W D})^{-1}* p \mathbf{W D}|| \leq ||\mathbf{I}|| + ||(\mathbf{I} - p \mathbf{W D})^{-1}* p \mathbf{W D}|| \leq ||\mathbf{I}|| + ||(\mathbf{I} - p \mathbf{W D})^{-1}||*|p|*||\mathbf{W D}|| = 1 +||(\mathbf{I} - p \mathbf{W D})^{-1}|| * p*1 \Longrightarrow ||(\mathbf{I} - p \mathbf{W D})^{-1}|| \leq  1 +||(\mathbf{I} - p \mathbf{W D})^{-1}|| * p \Longrightarrow ||(\mathbf{I} - p \mathbf{W D})^{-1}|| - ||(\mathbf{I} - p \mathbf{W D})^{-1}||*p \leq 1 \Longrightarrow ||(\mathbf{I} - p \mathbf{W D})^{-1}||*(1-p) \leq 1 \Longrightarrow ||(\mathbf{I} - p \mathbf{W D})^{-1}|| \leq 1/(1-p)$ 

Luego, 

$cond(\mathbf{I} - p \mathbf{W D}) = ||\mathbf{I} - p \mathbf{W D}|| * ||( \mathbf{I} - p \mathbf{W D})^{-1}|| \leq (1+p)/(1-p)$

Esto nos muestra que la condicion de la matriz $(\mathbf{I} - p \mathbf{W D})$ depende del valor de p, cuando p tiende a 0 la condicion de la matriz tiende a 1 por lo que estaria bien condicionada. Pero cuando p tiende a 1 lo unico que sabemos es que la condicion es menor a infinito por lo que perdemos informacion.
